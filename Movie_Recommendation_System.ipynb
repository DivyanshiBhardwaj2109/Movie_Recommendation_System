```python
df =df.dropna(subset=['Movie_ID', 'Movie_Title'])  
df= df.dropna(subset=['User_ID', 'Rating'])
```

```python
df = df[(df['Rating'] >= 1) & (df['Rating'] <= 5)]
```

```python
user_movie_matrix = df.pivot_table(index='User_ID', columns='Movie_ID', values='Rating').fillna(0)
```

```python
user_sim_matrix = cosine_similarity(user_movie_matrix)
user_sim_df = pd.DataFrame(user_sim_matrix, index=user_movie_matrix.index, columns=user_movie_matrix.index)
```

```python
target_user = 38
```

```python
sim_users = user_sim_df[target_user].sort_values(ascending=False)
sim_users = sim_users.drop(target_user)
```

```python
sim_users
```

```python
movies_seen = user_movie_matrix.loc[target_user][user_movie_matrix.loc[target_user] > 0].index.tolist()
```

```python
recommendations = []
for sim_user in sim_users.index:
    high_rated_movies = user_movie_matrix.loc[sim_user][user_movie_matrix.loc[sim_user] >= 4].index.tolist()
    for movie in high_rated_movies:
        if (movie not in movies_seen) and (movie not in recommendations):
            recommendations.append(movie)
    if len(recommendations) >= 5:
        break
```

```python
recommended_titles = df[df['Movie_ID'].isin(recommendations)][['Movie_ID', 'Movie_Title']]
print("Top recommended movies for user", target_user)
print(recommended_titles)
```

```python
print("Most similar users to", target_user)
print(sim_users.head(3)) 
```

```python
print(type(df))
print(type(user_movie_matrix))
```

```python
for movie in recommendations:
    liked_by = []
    # Check among top 10 similar users
    for sim_user in sim_users.index[:10]:
        Rating = user_movie_matrix.loc[sim_user, movie]
        if Rating >= 4:
            liked_by.append(sim_user)
    title = df[df['Movie_ID'] == movie]['Movie_Title'].values[0]
    print(f"Movie: {title} (ID: {movie}) liked by users: {liked_by}")
```

```python
user_movie_matrix.to_csv("user_movie_matrix.csv")
df.to_csv("movies.csv", index=False)
user_sim_df.to_csv("user_sim_df.csv")
```

```python

import joblib
import os

# Create folder to store model
os.makedirs("models", exist_ok=True)

# Save model components
model_data = {
    "user_similarity": user_sim_df,
    "user_movie_matrix": user_movie_matrix
}

joblib.dump(model_data, "movie_recommendation_model.pkl")

print("Model saved successfully!")
```
